```python
import numpy as np
from scipy.stats import *
import random
# for reproducability
np.random.seed(42)
```

```python
def birth_death_chain(L, alpha, beta, N):
    # list of times of each jump
    t_n_list = []

    for _ in range(N):
        state = 0
        t = 0.0

        while state < L:
            if state == 0:
                # reflecting state: only q(0,1)
                total_rate = alpha
                # sample time MC jumps to next state
                t_n = np.random.exponential(1.0 / total_rate)
                # move forward in chain
                state += 1
            else:
                total_rate = alpha + beta
                # sample time MC jumps to next state
                t_n = np.random.exponential(1.0 / total_rate)
                # if U(0,1)< alpha / (alpha + beta), move foward
                #  else move backward
                if np.random.uniform() < (alpha / total_rate):
                    state += 1
                else:
                    state -= 1
            t += t_n

        t_n_list.append(t)

    return np.array(t_n_list)
```

```python
L = 20       # size of chain
alpha = 1.0  # q(i,i+1)
beta = 1.0   # q(i,i-1)
```

N = 1000

```python
m_0 = birth_death_chain(L, alpha, beta, N=1000)

simulated_m_0 = np.mean(m_0)
theoretical_m_0 = L * (L + 1) / 2

simulated_var = np.var(m_0)

print(f"Simulated Mean: {np.round(simulated_m_0, 5)}")
print(f"Theoretical Mean: {np.round(theoretical_m_0, 5)}")
print(f"Simulated Variance: {np.round(simulated_var, 5)}")
```
