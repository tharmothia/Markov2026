```python
import numpy as np
```

```python
# rate customers leave the queue
def mu(n):
  return 60 + 5 * (n - 1)

# normalization coefficient
def normalization(n, lmbda):
  summands = []      # store computed terms in sum

  # compute each term in sum
  for x in range(0,n):
    summand = 1
    # compute mu_n
    # mu_n = mu_1 + mu_2 + ... + mu_n
    for i in range(1,x+1):
      summand *= 30 / mu(i)
    summands.append(summand)

  # return normalization coefficient of pi(0):
  #   1 / cumulative sum of terms computed
  return 1 / np.cumsum(summands)[-1], np.cumsum(summands)[-1]
```

```python
# probability that no customers are in line
#    approximate normalization coefficient by keeping many terms of the series
p_zero, rho = np.round(normalization(1000, 3), 5)

# probability that there are customers in line
p_some = 1 - p_zero

print(f'The probability that there are customers in line is {p_some}')

print(f'The expected number of ice creams sold per hour is {p_some * 60}')
```
