```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import factorial
from scipy.stats import *
import random
import plotly.express as px
```

# Part c

```python
def arrival_times(arrival_rate, max_time):
  current_time = 0.0
  arrival_times = []

  # generate events while t in [0, T]
  while current_time < max_time:
      # Sample the next time increment using expovariate
      inter_arrival = random.expovariate((arrival_rate))
      current_time += inter_arrival

      if current_time <= max_time:
          arrival_times.append(current_time)

  return arrival_times
```

```python
def rate(t):
  return 0.5 * (1 + (t / 30)**2)

# max rate (max t in [0,120])
max_rate = rate(120)
print(f'The max rate is {max_rate}\n')

# sample arrival times T1, T2, . . . according to a HPPP with rate λMAX
case_times = arrival_times(max_rate, 120)
print(f'The number of cases sampled is {len(case_times)}')
```

```python
keep_cases = []
# keep arrival time Ti with probability λ(Ti)/λMAX
# generate U ~ Uniform(0,1)
# if rate ratio >= U, keep case
for t in case_times:
  U = random.uniform(0,1)
  if U <= rate(t) / max_rate:
      keep_cases.append(t)

print(f'The number of cases kept is {len(keep_cases)}')
```

```python
# Plot a histogram of report times per day for days 0 to 120.
# Compare the total number of reports with the expected value calculated in (b).
fig = px.histogram(keep_cases, x = keep_cases, nbins = 120, range_x = [0,120],
                   labels = {
                       "x" : "Day"
                   },
                   title = 'Flu Cases Reported Per Day')
fig.show()
```
