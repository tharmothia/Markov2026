```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
m = np.arange(1, 100)

# E[tau_m]
def expected_value(m, beta):
  summands = []
  summand = 0

  for k in range(1, m):
    summand += 1 / (m * beta)
    summands.append(summand)

  return np.cumsum(summands)


# y(tau_m)
def exponential_growth(m, beta):
  return np.log(m) / beta


# to plot E[tau_m] and y(tau_m) together
def results(m, beta):
  # m is the max number of particles the Yule Process is simulated to
  # beta is birth rate

  # E[tau_m]
  expected_tau = expected_value(m, beta)

  # y(tau_m)
  exponential_tau = [exponential_growth_vec(k, beta) for k in range(1, m)]

  return expected_tau, exponential_tau
```
